```python
try:
    from sympy import nextprime, isprime
except ImportError:                      # not every notebook ships sympy
    !pip install --quiet sympy
    from sympy import nextprime, isprime

import secrets

def generate_params(bits=1024):
    """Generate Schnorr parameters: large prime p, prime q dividing p-1, and generator g of order q."""
    while True:
        # Generate a large prime q (for the subgroup order)
        # q should be roughly bits/2 long for security
        q = nextprime(secrets.randbits(bits // 2))

        # Candidate for p = 2q + 1 (Sophie Germain prime and safe prime construction)
        p = 2 * q + 1

        # Check if p is prime
        if isprime(p):
            break

    # Find a generator g of order q modulo p
    # In this p=2q+1 construction, any quadratic residue will have order q.
    g = 1
    while g == 1:
        h = secrets.randbelow(p - 2) + 2  # Pick h in [2, p-1]
        g = pow(h, 2, p) # g = h^2 mod p. This ensures g is a quadratic residue of order q.

    return p, q, g

def schnorr_prove(x, y, g, p, q, rounds=10):
    """Prover runs 'rounds' iterations of the Schnorr protocol.
    Returns a list of (r, c, s) for each round, simulating the verifier's challenge c.

    Args:
        x (int): The prover's private key.
        y (int): The prover's public key (y = g^x mod p).
        g (int): The generator of the cyclic group.
        p (int): The prime modulus.
        q (int): The prime order of the subgroup generated by g.
        rounds (int): The number of challenge-response rounds.

    Returns:
        list: A list of tuples, where each tuple contains (r, c, s) for a round.
    """
    proofs = []
    for _ in range(rounds):
        # 1. Prover chooses random nonce k in Z_q
        k = secrets.randbelow(q)

        # 2. Prover computes commitment R = g^k mod p
        r = pow(g, k, p)

        # 3. Simulate verifier generating a random challenge c in Z_q
        # In a real protocol, this challenge c would be sent by the verifier.
        c = secrets.randbelow(q)

        # 4. Prover computes response s = (k - x*c) mod q
        s = (k - (x * c)) % q

        proofs.append((r, c, s))
    return proofs

def schnorr_verify(y, g, p, q, r, c, s):
    """Verifier checks the proof (r, c, s).

    Args:
        y (int): The prover's public key.
        g (int): The generator.
        p (int): The prime modulus.
        q (int): The prime order of the subgroup.
        r (int): The prover's commitment.
        c (int): The challenge.
        s (int): The prover's response.

    Returns:
        bool: True if the proof is valid, False otherwise.
    """
    # Check if s and c are within the valid range Z_q
    if not (0 <= s < q and 0 <= c < q):
        return False

    # Verifier computes R_prime = (g^s * y^c) mod p
    r_prime = (pow(g, s, p) * pow(y, c, p)) % p

    # Verifier checks if R == R_prime
    return r == r_prime

def schnorr_cheat(y, g, p, q, rounds=10):
    """Simulate a prover who doesn't know x.
    They must guess challenges in advance to try and create a valid-looking proof.

    Args:
        y (int): The target public key.
        g (int): The generator.
        p (int): The prime modulus.
        q (int): The prime order of the subgroup.
        rounds (int): The number of rounds to simulate.

    Returns:
        list: A list of tuples, where each tuple contains (r, c_guessed, s) for a round.
    """
    cheat_proofs = []
    for _ in range(rounds):
        # 1. Cheater chooses a random challenge c_prime they hope to receive from the verifier
        c_prime = secrets.randbelow(q)

        # 2. Cheater chooses a random response s
        s = secrets.randbelow(q)

        # 3. Cheater computes R = (g^s * y^c_prime) mod p
        r = (pow(g, s, p) * pow(y, c_prime, p)) % p

        cheat_proofs.append((r, c_prime, s))
    return cheat_proofs
```

## Schnorr Protocol Demonstration

This section demonstrates the Schnorr identification protocol, including how an honest prover generates a proof and how a malicious prover (cheater) attempts to forge one. For simplicity in this implementation, the `schnorr_prove` function simulates the verifier's challenge generation. In a real zero-knowledge proof, the verifier would generate and send the challenge `c`.

```python
# 1. Generate System Parameters (p, q, g)
print("Generating Schnorr parameters...")
p, q, g = generate_params(bits=256) # Use smaller bits for faster demonstration

print(f"p = {p}\nq = {q}\ng = {g}\n")

# 2. Prover generates a private key (x) and public key (y)
x = secrets.randbelow(q) # Private key x in Z_q
y = pow(g, x, p)         # Public key y = g^x mod p

print(f"Prover's Private Key (x): {x}")
print(f"Prover's Public Key (y): {y}\n")

# 3. Honest Prover runs the protocol
print("--- Honest Prover's Proof ---")
honest_proofs = schnorr_prove(x, y, g, p, q, rounds=1)

for i, (r, c, s) in enumerate(honest_proofs):
    print(f"Round {i+1}: r={r}, c={c}, s={s}")

    # Verifier checks the proof
    is_valid = schnorr_verify(y, g, p, q, r, c, s)
    print(f"  Verifier says: {is_valid} (Expected: True)\n")

# 4. Malicious Prover (Cheater) tries to forge a proof without knowing x
print("--- Malicious Prover's Attempt ---")
# The cheater will guess challenges in advance
cheater_proofs = schnorr_cheat(y, g, p, q, rounds=1) # One round for simple demo

for i, (r_cheat, c_guessed, s_cheat) in enumerate(cheater_proofs):
    print(f"Round {i+1} (Cheater): r={r_cheat}, c_guessed={c_guessed}, s={s_cheat}")

    # In a real protocol, the verifier generates a *new* random challenge c_actual.
    # The cheater's forged proof will only pass if c_actual == c_guessed.
    # For this demonstration, we'll use a new random challenge from the 'verifier'.
    c_actual_verifier = secrets.randbelow(q)

    print(f"  Verifier's actual challenge: {c_actual_verifier}")

    # The verifier would then check the proof using r_cheat, c_actual_verifier, and s_cheat
    # This should fail with very high probability if c_actual_verifier != c_guessed
    is_valid_cheat = schnorr_verify(y, g, p, q, r_cheat, c_actual_verifier, s_cheat)
    print(f"  Verifier says (with actual challenge): {is_valid_cheat} (Expected: False if c_actual_verifier != c_guessed)\n")

    if c_actual_verifier == c_guessed:
        print("  NOTE: Cheater's guessed challenge matched verifier's actual challenge! This is rare but possible.")
        print(f"  Verifier says (with actual challenge, which matched guess): {is_valid_cheat} (Expected: True in this rare case)\n")

# To truly demonstrate cheating, one would typically run many rounds and observe the low probability of success.
```

# Cryptographic System Migration Plan

## 1. The Problem

*   **Centralized Single Point of Failure**: Traditional key management might rely on a single custodian or a limited number of highly privileged individuals, making the system vulnerable to compromise if that single point is breached or becomes unavailable. Similarly, centralized authentication mechanisms can be attractive targets for attackers.
*   **Outdated Algorithms**: Older cryptographic algorithms may be susceptible to modern attack techniques (e.g., brute-force attacks against smaller key sizes, known vulnerabilities in hashing functions).
*   **Lack of Auditability and Transparency**: Legacy systems might lack robust logging and audit trails, making it difficult to track key usage, access, or authentication attempts, which hinders compliance and incident response.
*   **Scalability Challenges**: Centralized key stores or authentication servers can become bottlenecks as the organization grows, leading to performance issues.
*   **Poor User Experience**: Complex or cumbersome authentication processes can lead to user frustration and reliance on insecure workarounds.

Our current system, for instance, uses a single master key stored in an HSM for critical data encryption, and password-based authentication for internal services, relying on SHA-1 hashing for password storage. This setup is a prime candidate for a security and operational overhaul.

## 2. Proposed Migration Mechanism

We propose migrating to a decentralized and threshold-based cryptographic system leveraging principles demonstrated by Shamir's Secret Sharing (SSS) for key management and Schnorr protocols for decentralized authentication.

### Phase 1: Key Management Migration using Shamir's Secret Sharing

1.  **Decomposition of Master Key**: The existing master key will be decomposed using Shamir's Secret Sharing into `n` shares, requiring `t` shares to reconstruct the original key (`t < n`). For example, `n=5` shares with a `t=3` threshold.
2.  **Distributed Storage**: These `n` shares will be securely distributed among different custodians (e.g., department heads, independent security personnel) and stored in geographically diverse, secure locations (e.g., different HSMs, secure vaults, encrypted hardware).
3.  **Policy Enforcement**: Access control policies will be implemented to ensure that at least `t` custodians must concur to reconstruct the key for any operation requiring the master key (e.g., disaster recovery, system upgrades). This eliminates the single point of failure.

### Phase 2: Authentication Migration using Schnorr Protocol

1.  **Generate Schnorr Parameters**: System-wide Schnorr parameters (`p`, `q`, `g`) will be generated, where `q` is a large prime, `p` is a prime modulus, and `g` is a generator of a subgroup of order `q` modulo `p`.
2.  **User Key Pair Generation**: Each user will generate their own Schnorr private key (`x`) and corresponding public key (`y = g^x mod p`). The private key `x` will be stored securely on the user's device (e.g., TPM, secure enclave), possibly protected by a passphrase. The public key `y` will be registered with the authentication service.
3.  **Challenge-Response Authentication**: During authentication, the user's client will engage in a Schnorr zero-knowledge proof with the server:
    *   Client selects a random `k` and computes `R = g^k mod p`.
    *   Client sends `R` to the server.
    *   Server generates a random challenge `c` and sends it to the client.
    *   Client computes `s = (k - x*c) mod q` and sends `s` to the server.
    *   Server verifies `R == (g^s * y^c) mod p`.
4.  **Decentralized Verification**: This process ensures that the user proves knowledge of `x` without ever revealing `x` to the server, enhancing privacy and security against server-side breaches. The authentication logic can be distributed across multiple independent verifiers.

## 3. Trade-offs (size, cost, governance)

*   **Size**:
    *   **Pros**: Shamir's Secret Sharing increases resilience by distributing the secret, but individual shares can be larger than the original secret. Schnorr public keys are small (e.g., 256 bits).
    *   **Cons**: Storing multiple shares and managing their secure distribution adds to storage overhead compared to a single centralized key. Schnorr proofs (r, c, s) are larger than simple password hashes.
*   **Cost**:
    *   **Pros**: Enhanced security can reduce the cost of breaches. Over time, decentralized systems can offer better scalability than continuously upgrading monolithic, centralized infrastructure.
    *   **Cons**: Initial setup and implementation of SSS and Schnorr protocols require significant development effort, cryptographic expertise, and potentially new hardware (e.g., more HSMs for share storage, secure elements on user devices). Training custodians for SSS and users for new authentication flows also incurs costs.
*   **Governance**:
    *   **Pros**: SSS enforces a robust multi-party control mechanism, dramatically improving governance over critical keys by requiring consensus. Schnorr authentication decentralizes trust away from a single authentication server.
    *   **Cons**: Increased complexity in key recovery procedures (coordinating `t` parties). Managing user Schnorr key pairs and their secure storage on individual devices adds an administrative burden. The introduction of new cryptographic primitives requires careful policy updates and auditing frameworks.

## 4. User Impact

*   **Key Management (SSS)**:
    *   **Reduced Risk**: Users benefit from a significantly reduced risk of a single point of failure compromising the master key, leading to higher data security.
    *   **Increased Complexity (Custodians)**: Key custodians will have a more involved process for key reconstruction, requiring coordination and adherence to strict protocols.
*   **Authentication (Schnorr)**:
    *   **Enhanced Security**: Users' private keys are never transmitted, making their authentication credentials highly resistant to phishing, man-in-the-middle attacks, and server breaches.
    *   **Improved Privacy**: The zero-knowledge nature means no sensitive information (like passwords) leaves the user's device during authentication.
    *   **Potential for Friction**: Initial setup of Schnorr key pairs on devices might be unfamiliar. Relying on secure elements could limit device flexibility. However, well-designed user interfaces can mitigate this.
    *   **Passwordless Future**: This migration paves the way for a truly passwordless authentication experience, leveraging biometrics or device-based security.
